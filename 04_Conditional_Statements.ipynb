# Conditional Statement
A conditional statement in Python is used to make decisions in a program. It executes a block of code based on whether a condition is `True` or `False` .



- Types of conditional statements:
> 1. if — checks a condition.
> 2. if-else — chooses between two outcomes.
> 3. if-elif-else — checks multiple conditions.
> 4. Nested if — an if statement inside another if statement.

<b>`Syntax of a Conditional Statement`</b>


```python
if condition:
    # Code executes if condition is True
elif another_condition:
    # Code executes if another condition is True
else:
    # Code executes if all conditions are False
```
