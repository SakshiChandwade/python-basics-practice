# File I/O / File Handling 
- File handling in Python involves managing data flow between your program and the storage device by opening, reading, writing, and closing files.
- File I/O (Input/Output), also known as file handling, refers to the process of reading data from files and writing data to files within a program.
- It is essential for managing data stored in external files.
- Storing Large Datasets
- Data Presistencs
- Data interchanges

**Modes (text or binary):**
- `r` – Read (default)
- `w` – Write (creates/truncates file)
- `a` – Append
- `x` – Exclusive creation
- `t` – Text mode (default)
- `b` – Binary mode
- `+` – Read/write update
  
**Reading Files**
- `f.read()` – Read entire file
- `f.readline()` – Read one line
- `f.readlines()` – Read all lines into a list

**Writing Files** 
- Syntax : `f.write("text")`

**Closing Files**
- Syntax : `f.close()`
  
**Deleting Files**
- Syntax : import os
- os.remove("filename.txt")   # delete file
- os.rmdir("foldername")      # delete folder

**File Object Methods** 
Common methods include:
- `close()`
- `read()`
- `write()`
- `seek()`
- `tell()`
- `truncate()`
- `writelines()`

Syntax :
with open("filename.txt", "mode") as file_variable:
    # Perform file actions (read, write, etc.)
    # The file is automatically closed outside this block

**Program 1: Writing to a File ("w")**

```python
# Writing content to a text file
with open("demo_File_Handling.txt", "w") as file:
    file.write("Hello World!\n")
    file.write("Python file handling is straightforward.\n")
print("File written successfully!")
```

**Program 2:  Reading from a File ("r")**

```python
# Approach 1: Reading the entire file content at once
with open("demo_File_Handling.txt", "r") as file:
    content = file.read()
    print("--- Full Content ---")
    print(content)

# Approach 2: Reading line-by-line using a loop (Best for large files)
with open("demo_File_Handling.txt", "r") as file:
    print("--- Line-by-Line Iteration ---")
    for line in file:
        print(line.strip())  # .strip() removes trailing extra newlines (\n)
```

**Program 3: Appending Data ("a")** Use this mode to add new information without wiping out what is already in your file.

```python
# Appending text to an existing file
with open("demo_File_Handling.txt", "a") as file:
    file.write("Adding a new line without deleting previous text in file handling a Appending Data.\n")
```
