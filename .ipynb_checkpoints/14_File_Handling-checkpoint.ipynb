# File I/O / File Handling 
- File handling in Python involves managing data flow between your program and the storage device by opening, reading, writing, and closing files.
- File I/O (Input/Output), also known as file handling, refers to the process of reading data from files and writing data to files within a program.
- It is essential for managing data stored in external files.
- Storing Large Datasets
- Data Presistencs
- Data interchanges

**Modes (text or binary):**
- `r` – Read (default)
- `w` – Write (creates/truncates file)
- `a` – Append
- `x` – Exclusive creation
- `t` – Text mode (default)
- `b` – Binary mode
- `+` – Read/write update
  
**Reading Files**
- `f.read()` – Read entire file
- `f.readline()` – Read one line
- `f.readlines()` – Read all lines into a list

**Writing Files** 
- Syntax : `f.write("text")`

**Closing Files**
- Syntax : `f.close()`
  
**Deleting Files**
- Syntax : import os
- os.remove("filename.txt")   # delete file
- os.rmdir("foldername")      # delete folder

**File Object Methods** 
Common methods include:
- `close()`
- `read()`
- `write()`
- `seek()`
- `tell()`
- `truncate()`
- `writelines()`

**Syntax :**
with open("filename.txt", "mode") as file_variable:
    # Perform file actions (read, write, etc.)
    # The file is automatically closed outside this block

###  Programs : write a  Second way of file handling programs 
`With Syntax`

**Program 1:**  Read File 

```python
with open("demo.txt","r") as f:    # withSyntax as means alias i.e, as this shortcut name 
    data=f.read()
    print(data)
```

**Program :** Read line by line `with`

```python
with open("demo.txt","r") as f:
    lines =f.readlines()
    for line in lines :
        print (line.strip())
```

**Program 2: Writing to a File ("w")** `with` syntax 

```python
# Writing content to a text file
with open("demo_File_Handling.txt", "w") as file:
    file.write("Hello World!\n")
    file.write("Python file handling is straightforward.\n")
print("File written successfully!")
```

**Program 3:`with` syntax Appending Data ("a")** Use this mode to add new information without wiping out what is already in your file.

```python
# Appending text to an existing file
with open("demo_File_Handling.txt", "a") as file:
    file.write("Adding a new line without deleting previous text in file handling a Appending Data.\n")
```

```python
import csv
data=[
    {"Name": "Raju","Age":30},
    {"Name":"Ramu","Age":25},
    {"Name":"Sanu","Age":35},
    {"Name":"Pappu","Age":22}
]
with open ("input.csv","w",newline = "") as file:
    fieldnames = ["Name","Age"]
    writer = csv.DictWriter(file,fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(data)
```

```python
# Open input and output files
with open("input.csv","r") as infile,open ("output.csv","w",newline="") as outfile:
# Create a DictReader object to read the input CSV
    reader = csv.DictReader(infile)

# Create a DictWriter object to write to the output CSV
# Ensure fieldnames are specified for DictWriter
    writer = csv.DictWriter(outfile,fieldnames = reader.fieldnames)

# Write the header row to the output file
    writer.writeheader()

# Iterate over each row in the input CSV
    for row in reader : 
        if int (row['Age'])>25:
            writer.writerow(row)
```
