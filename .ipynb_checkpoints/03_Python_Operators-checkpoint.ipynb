# Type conversion / Type Casting 
- Type conversion in Python is the process of changing a value from <b> one data type to another data type</b>.
- Python supports two types of type conversion:
  > Implicit Conversion &
  > Explicit Conversion.


<B>Program 1 : By defalut set String</b>

```python
num = 10
num_str = str(num)
print(num_str,type(num_str))
```

```python
print (num,type(num))
```

<B>Program 2 : converting string into integer </b>

```python
num = 10
num_str = int(num)
print(num,type(num))
```

<b> Program 3 : converting string into integer  </b>

```python
num = int ((input("Enter Number:")))
num=num+3
print (num)
print (type(num))
```

<b>Program 4: converting into float into integer</b>

```python
pi = 3.14
int(pi)
```

<b>Program 5 : Taking input from user as float (int into float)</b>

```python
num = float ((input("Enter number: ")))
num = num+2
print (num)
print (type(num))
```

<b>Program 6: int to float </b>

```python
num = 100
num_float = float(num)
print(num_float,type(num_float))
```

<b>Program 7 : bool to int</b>

```python
bool_value = True
int(bool_value)
```

# Comparision Operations 

```python
x = 5
y = 10
print (x==y)
print (x>y)
print(x<y)
print(x>=y)
print(x<=y)
print(x != y )
```

# Logical Operations

```python
a = True
b = False
print (a and b) # AND
print (a or b) # OR
print (not a) # NOT
```
