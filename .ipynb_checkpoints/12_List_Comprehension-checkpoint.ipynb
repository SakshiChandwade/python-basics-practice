# List Comprehension 
List comprehension is a concise way to create new lists by applying an expression to each item in an existing iterable like a ( list, tuple or range).

<b> Syntax for list comprehension </b>
> `Syntax`: [ expression <b>for</b> item <b>in</b> iterable <b>if</b> condition]

<b>Program 1 :</b> 

```python
numbers = [1,2,3,4,5]
squares = [x ** 2 for x in numbers]
print (squares)
```

<b>Program 2 :</b> 

```python
words = ['hello','world','python']
[ word.upper() for word in words]
```

### List Comprehension with conditions

<b>Program 1 : </b>

```python
numbers = [1,2,3,4,5,6,7]
even = [x for x in numbers if x % 2 == 0]
print(even)
```

<b>Program 2 : Extract Digits From String </b>

```python
text = "abc123def456"
digits = [char for char in text if char.isdigit()]
print(digits)
```

### Nested List Comprehensions 
- Nested list comprehensions involve using multiple for loops within a single list comprehension.
- Nested list comprehension is to flatten a matrix, which means converting a nested list structure into a single, one-dimensional list.
- `Syntax`: [expression for outer_item in outer_iterable for inner_item in inner_iterable]. 



<b>Program 1 : Flatten </b>

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
flatten = [element for row in matrix for element in row ]
print (flatten)
```

<b>Program 2 : Transposed</b>

```python
matrix
```

```python
transposed = [[row[i]for row in matrix] for i in range (len(matrix[0]))]
print (transposed)
```

### Real Life Usecase

```python
emails = ['Sakshi@gmail.com','Raju@yahoo.co.in','mahii11@outlook.com']
domains = [i.split('@')[1]for i in emails]
print (domains)
```
