# Introduction to Pandas 
- Pandas is an open-source Python library used for data manipulation, data analysis, and data processing
- It provides powerful and flexible data structures to work with structured data such as tables, spreadsheets, CSV files, and databases.

**Features of Pandas**
1. **Data Handling:** It can handle structured data in different formats.
2. **Data Cleaning:** It helps to identify and remove missing or duplicate values.
3. **Data Analysis:** It provides functions for statistical analysis.
4. **Data Filtering:** It allows users to select and filter data based on conditions.
5. **Data Integration:** It supports merging, joining, and concatenating datasets.
6. **File Handling:** It can read and write CSV, Excel, JSON, and SQL data.
7. **Data Visualization:** It integrates with Matplotlib for creating graphs and charts.
8. **Time Series Analysis:** It provides tools to work with dates and time-based data.


 ### 1. Key Data Structures:
- **Series:** A 1-dimensional labeled array that holds data of any type (integers, strings, floats). Think of it as a single column in an Excel spreadsheet.
- **DataFrame:** A 2-dimensional, size-mutable table with labeled rows and columns. Think of it as an entire spreadsheet or a SQL database table.
> - **Inshort : Key Data Structures:**
> - **Series** = one list
> - **DataFrame** = full table

### 2. Why Pandas is Important in Data Science
- **Handles Real-World Messy Data:** Machine learning models fail if given blank or corrupt values. Pandas provides single-line tools to clean missing entries, fix formatting, and drop duplicates.
- **High Speed (Vectorization):** Built on C-optimized NumPy arrays, Pandas performs math across millions of rows simultaneously without slow Python for loops.
- **Universal File Bridge:** It seamlessly reads and writes multiple formats (read_csv(), read_excel(), read_json(), read_sql()) in one line of code.

### Series Indexing
- Indexing means accessing a particular value from a Series using its index.
- There are two common ways to access values.
> 1. Positive Position Indexing: `series.iloc[position]` 
> 2. Label-Based Indexing: `series.loc[label]`

```python
# Example 1: create series using list[] in data structure  with Series Indexing 
sales = pd.Series([2,3,5])  

#Series Indexing [2]
sales[2]
```

```python
# Example 2 : Indexing Pandas Series 
data = pd.Series([10,20,30,40,20])

# accessing element starting = 2 till the end  
data[2:]
```

```python
# Example 3: Indexing Explicitly
pd.Series([0,1,2,3],index = ['a','b','c','d'])
```

```python
# Example  4: Index as a Sequences Or use Functions to specify the index 
import numpy as np
pd.Series(np.array(range(0,10))**2,index= range(0,10))
```

```python
# Example : Positive Position Indexing 
s = pd.Series([10, 20, 30, 40, 50])

# Syntax : series.iloc[position]
print(s.iloc[0]) 
print(s.iloc[2])
```

```python
# Example : Label-Based Indexing
# Syntax : series.loc[label]

s = pd.Series([85, 90, 78],
    index=['A', 'B', 'C'])
print(s.loc['A'])
print(s.loc['C'])
```

### Series Slicing
Slicing means extracting multiple values from a Series using a range of indexes.
1. Basic Slicing
2. Slicing with Custom Labels
3. Slicing from Beginning
4. Slicing till End
5. Reverse Slicing

```python
# 1. Basic Slicing 
# Example : 
s = pd.Series([10, 20, 30, 40, 50])
# Syntax : series[start:end]
print(s.iloc[1:4])
# Ending position is excluded in iloc
```

```python
# 2. Slicing with Custom Labels 
s = pd.Series(
    [85, 90, 78, 95, 88],
    index=['A', 'B', 'C', 'D', 'E']
)

print(s.loc['B':'D'])
```

```python
# 3.Slicing from Beginning
print(s.iloc[:3])
```

```python
# 4. Slicing till End : Returns values starting from position 2 till the end.
print(s.iloc[2:])
```

```python
# 5. Reverse Slicing  : Displays the Series in reverse order.
print(s.iloc[::-1]) 
```

### Pandas Series Attributes
Attributes provide information about a Series. They describe the data but generally do not modify it.
- `s.index`	Shows index labels	**s.index**
- `s.values`	Shows values	**s.values**
- `s.dtype`	Shows data type	**s.dtype**
- `s.shape`	Shows dimensions	**s.shape**
- `s.size`	Total number of elements	**s.size**
- `s.ndim`	Number of dimensions	**s.ndim**
- `s.name`	Series name	**s.name**
- `s.empty`	Checks whether Series is empty	**s.empty**
- `s.nunique()`	Number of unique values	**s.nunique()**

```python
#Example : Pandas Series Attributes 
print("Index:", s.index)
print("Values:", s.values)
print("Data Type:", s.dtype)
print("Shape:", s.shape)
print("Size:", s.size)
print("Dimension:", s.ndim)
```

### Methods Pandas Series 
Methods are functions used to perform specific operations on a Series, such as calculations, sorting and data analysis.
1. **Basic Methods**
- `s.head()` : Displays first 5 values
- `s.tail()` : Displays last 5 values
- `s.count()`: Counts non-missing values
- `s.unique()`: Returns unique values
- `s.nunique()`: Counts unique values
- `s.value_counts()`: Counts frequency of each value
- `s.to_list()`: Converts Series into a list
2. **Statistical Methods**  : These methods are very useful in Data Science and Data Analytics.
- `s.sum()`: Calculates total
- `s.mean()`: Calculates average
- `s.median()`: Finds middle value
- `s.mode()`: Finds most frequent value
- `s.min()`:  Finds minimum
- `s.max()`:  Finds maximum
- `s.std()`:  Standard deviation
- `s.var()`:  Variance
- `s.describe()`: Statistical summary

3.**Sorting Methods**:
- `s.sort_values()`: Sort values in ascending order
- `s.sort_values(ascending=False)`: Sort values in descending order
- `s.sort_index()`: Sort index
- `s.reset_index()`: Reset index

```python
# Example : Methods pandas series , use this dataset
s = pd.Series([10, 20, 30, 40, 50, 20])
print (type(s))
print(s)
```

```python
#Example : s.head() 
print(s.head())
```

```python
#Example: s.tail()
print (s.unique())
```

```python
# Example: s.nunique()
print(s.nunique())
```

```python
#Example:s.value_Counts()
print(s.value_counts())
```

```python
#Example : Statistical Methods 
print("Total:", s.sum())
print("Average:", s.mean())
print("Maximum:", s.max())
print("Minimum:", s.min())
print("Median:", s.median())
```

```python
#Example: Sorting Methods(Sort values in ascending order)
s.sort_values()
```

```python
#Example: Sorting Methods(Sort values in descending order)
s.sort_values(ascending=False)
```

```python
#Example: Sorting Method(Sort index)
s.sort_index()
```

```python
#Example: Reset index
s.reset_index()
```

### Pandas Series Operations
Operations are used to perform mathematical or logical calculations on Series values
1. **Arithmetic Operations**
- Addition
- Subtraction (-)
- Multiplication(*)
- Division(/)
- Modulus(%)
- Power(**)

2. **Comparison Operations**: Comparison operations return True or False.
3. **Logical Operations**: We can combine multiple conditions
- `AND(&)`:  (both conditions must be true).
- `OR(|)`: (at least one condition must be true).
- `NOT(~)` : (reverses the condition).

```python
#Example: Let's create two Series:
s1 = pd.Series([10, 20, 30, 40])
s2 = pd.Series([1, 2, 3, 4])
```

```python
#Example: Arithmetic Operations 
print(s1 + s2)
print(s1 - s2)
print(s1 * s2)
print(s1 / s2)
```

```python
#Example: Comparison Operations,Comparison operations return True or False.
s = pd.Series([10, 20, 30, 40, 50])

print(s > 25)
print(s == 20)
print(s < 30)
```

```python
#Example:Logical Operations 
s = pd.Series([10, 20, 30, 40, 50])

print(s[(s > 20) & (s < 50)])
```
