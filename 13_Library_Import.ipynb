# Library Import 1
- <B>LIBRARY:</B> A Python library is a group of modules packaged together to provide ready‑made functionality for specific tasks
- Library is a group of modules that contain functions, classes and methods to perform common tasks like data manipulation, math operations, web scraping and more.
- Python libraries make coding faster, cleaner and more efficient by providing ready-to-use solutions for different domains such as data science, web development, machine learning and automation.
- <b>Types of Python Libraries </b>
1. <b> Built‑in (Standard Library) :</b>
     > - These come with every Python installation and require no separate installation.
     > - It is a collection of modules that come bundled with every Python installation, we don’t need to install anything separately.
     > - <b>Examples of built-in modules: </b>
     >> - `math` : Mathematical operations
     >> - `os ` : Interact with the operating system
     >> - `datetime` : Date and time operations
     >> - `random` : Generate random numbers
     >> - `json` : Handles JSON data encoding and decoding.
     

2. <b> External (Third‑Party) Libraries :</b>
   > - These must be installed using tools like pip.
   > - External (third-party) libraries are not included with Python by default. You can install them easily using the pip package manager.
   >> - ` NumPy` : A library for **numerical computing** that provides support for arrays, matrices, and mathematical operations.
   >> - `Pandas` : A library for data analysis** and** manipulation using powerful data structures like **DataFrame** and **Series.**
   >> - `Matplotlib`: It is a **data visualization** library that helps you create a wide variety of static, animated and interactive plots. It supports charts such as bar graphs, line charts, histograms and scatter plots.
   >> - `SciPy`: A library for advanced scientific and mathematical computations built on top of NumPy.
   >> - `TensorFlow`: It is an open-source **machine learning and deep learning** framework developed by Google. It allows developers to build and train neural networks for tasks such as image recognition, natural language processing and predictive modeling.
   >> - `Scikit-learn`: It is a popular **machine learning** library built on top of NumPy and SciPy. It offers simple and efficient tools for **classification, regression, clustering and dimensionality reduction.**
   >> - `Scrapy`: A library for** web scraping** and extracting structured data from websites.
   >> - `PyTorch`:** A deep learning framework** with GPU support for building and training **neural networks.**
   >> - `PyGame`: It is a cross-platform library used for **developing 2D games** and multimedia applications. It provides modules for handling **graphics, sound and user input with ease.**
   >> - `Keras`: A high-level deep learning library that provides an easy interface for building neural networks.

- <b>Program of  `Math`  Library with Function</b>

<b> `date time` </b>

```python
# Syntax :   from module_name import function_name

# Example 1: 

from datetime import datetime
now =datetime.now()
print(now)
```

```python
formatted_date = now.strftime("%d-%m-%y %H:%M:%S")
print(formatted_date)
```

<b>Date Arithmetic:</b> The timedelta object, also from the datetime library, enables you to perform calculations with dates.

```python
from datetime import timedelta
future_date = now + timedelta(days=10)
print(future_date)
```
