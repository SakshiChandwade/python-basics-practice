# TUPLE () 
- A tuple in Python is a built-in collection data type used to store multiple items in a single variable
- <B>Tuples are defined using parentheses ()</B> with items separated by <B>commas</B>, or simply by <B>separating values with commas</B>.
- <B> Paranthesis Optional () in Tuple</b>

## Key Characteristics
- <B>Ordered: </B>Elements have a defined order that will not change.
- <B>Immutable: </B>You cannot modify, append, or delete items after creation.
- <B>Heterogeneous:</B> Can hold elements of different data types (e.g., strings, integers, lists) simultaneously.
